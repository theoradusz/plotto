# plotto showcase

The examples of `gallery.py`. A plot fits the cell and has copy and download buttons. `.interactive()` adds zoom, pan, hover and the legend toggle; a click pins the hover label.

```python
import gallery as g
from plotto import plot
```

## Write a plot

```python
p = (
    plot(g.uv254(), x="time", y="value", color="sensor")
    .line()
    .hline(y=5, color="red", dash="dashed")
    .labels(title="UV254 over one week", x=None, y="UV254 (1/m)")
)
p
```

```python
p.interactive()
```

## Limits, style, palette

```python
p.y(limits=[0, None]).theme("wota").color(palette="aizome")
```
